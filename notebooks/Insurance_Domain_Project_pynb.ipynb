# Data Cleaning

```python
## Dropping irrelevant columns
df = df.drop(columns= ['UPR FM','UPR TO'])
df = df.drop(columns= ['ENDORSEMENT DATE'])
df =df.drop(columns=['ENTRY COMPLETE DT'])
```

```python
import pandas as pd

# converting date columns to datetime
df ['POLICY FROM DATE'] = pd.to_datetime(df['POLICY FROM DATE'], format='%d/%m/%Y')
df['POLICY TO DATE'] = pd.to_datetime(df['POLICY TO DATE'], format='%d/%m/%Y')
df['APPROVAL DATE'] = pd.to_datetime(df['APPROVAL DATE'], format = '%d/%m/%Y')
```

## 2. Exploratory Analysis

### 1. How many policies per branch
### 2. How many policies Per product line?
### 3. What does total gross premium look like by branch and by product?
### 4. What's the overall claims frequency (% of policies with a claim > 0) and average loss ratio, split by product line?
### 5. New vs renewal business — what's the mix, and does loss ratio differ between the two?
### 6. Any branch or product that stands out as unusually high loss ratio compared to the rest?






```python
#1. How many policies per branch
df_clean.groupby('BRANCH CODE')['POLICY FROM DATE'].count()
```

```python
#2 How many policies per product line
df_clean.groupby('PRODUCT DESC')['POLICY FROM DATE'].count()
```

```python
# 3. What does total gross premium look like by branch and by product?
df_clean.groupby(['BRANCH CODE', 'PRODUCT DESC'])['GROSS PREMIU'].sum()
```

```python
# 4 What's the overall claims frequency (% of policies with a claim > 0) and average loss ratio, split by product line?
# logic - first create  flag/helper column where 1 represents a policy with a claim while 0 represents a policy without a claim

df_clean['Has_Claim'] = df_clean['CLAIM YTD PAID AMT'] > 0
```

```python
df_clean['Has_Claim'] = df_clean['Has_Claim'].astype(int)
```

```python
df_clean['Has_Claim'].value_counts()
```

```python
df_clean.groupby('PRODUCT DESC')[['Has_Claim', 'loss_ratio']].mean().sort_values('loss_ratio', ascending=False)
```

### Engineering and industrial fire products show the highest average loss ratios, consistent with their exposure to large, infrequent losses, while motor and liability lines show more moderate but frequent claims" — a defensible, industry-plausible sentence without overclaiming precision from what's ultimately a synthetic sample.

```python
# 5. New vs renewal business — what's the mix, and does loss ratio differ between the two?
df_clean.groupby('BUSINESS CLASS').agg(

    {
        'loss_ratio' : 'mean',
        'POLICY FROM DATE': 'count'

    }
)
```

### From the exploratory analysis done specifically from the new vs renewal business done, it is worth noting that, the mix: 3051 New Business vs 2949 Renewal Business — roughly a 51%/49% split, so close to evenly balanced. Loss ratio comparison: New Business (0.0086) is only marginally higher than Renewal (0.0074), interesting observation that new business tends to run slightly higher loss ratios than renewals
